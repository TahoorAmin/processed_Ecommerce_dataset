```python
import pandas as pd
orders_df=pd.read_csv("Day9_Orders.csv")
products_df=pd.read_csv("Day9_Products.csv")
customers_df=pd.read_csv("Day9_Customers.csv")
```

# Inspect the Datasets:
In this step, the first few rows of the Orders, Customers, and Products datasets are displayed to understand their structure and identify the related columns needed for merging.

```python
orders_df.head(3)
```

```python
products_df.head(3)
```

```python
customers_df.head(3)
```

# Examine Dataset Structure:
The column names and data types of each dataset are examined to understand the available information and identify suitable keys for combining the datasets.

```python
print("Order columns :")
print(orders_df.columns)

print("\nCustomer columns :")
print(customers_df.columns)

print("\nProduct columns :")
print(products_df.columns)
```

# Demonstrate DataFrame Concatenation:
The Orders dataset is divided into two parts and combined again using concat() to demonstrate row-wise DataFrame concatenation.

```python
orders_part1 = orders_df.iloc[:len(orders_df)//2]
orders_part2 = orders_df.iloc[len(orders_df)//2:]

orders_concat = pd.concat([orders_part1, orders_part2], ignore_index=True)

orders_concat.head()
```

# Identify Relationships Between Datasets:
The datasets are related through common identifiers. The Orders dataset contains Customer ID and Product ID, which can be used to connect it with the Customers and Products datasets. A left merge will be used to retain all order records while adding the corresponding customer and product information.

```python
order_customer=pd.merge(orders_df,customers_df,on="Customer_ID",how="left")
```

```python
order_customer.head(3)
```

```python
final_merged=pd.merge(order_customer,products_df,on="Product_ID", how="left")
```

```python
final_merged.head(3)
```

# Create Total Price:
A new Total Price column is created using the Quantity and Unit Price columns. The apply() function is used to calculate the total price for each order.

```python
def total_price(row):
    return row["Quantity"]*row["Unit_Price"]
final_merged["Total Price"]=final_merged.apply(total_price,axis=1)
```

```python
final_merged.head(3)
```

# Process Order Date:
The Order Date column is converted to datetime format to enable date-based operations and extraction of useful information such as month, day, and day of the week.

```python
final_merged["Order_Date"]=pd.to_datetime(final_merged["Order_Date"])
```

```python
final_merged["Order Year"]=final_merged['Order_Date'].dt.year
```

```python
final_merged["Order Month"]=final_merged['Order_Date'].dt.month
```

```python
final_merged["Order Day"]=final_merged['Order_Date'].dt.day
```

```python
final_merged["Order Day Of Week"]=final_merged['Order_Date'].dt.day_name()
```

```python
final_merged.head(3)
```

# Organize the Final Dataset:
The processed dataset is organized into a clear and meaningful column order by grouping order, customer, product, and derived date information together.

```python
final_merged = final_merged.loc[:, [
    "Order_ID", "Order_Date", "Customer_ID", "Product_ID",
    "Customer_Name", "City", "Region", "Membership_Type",
    "Product_Name", "Category", "Brand", "Unit_Price",
    "Quantity", "Total Price", "Payment_Method", "Order_Status",
    "Order Year", "Order Month", "Order Day", "Order Day Of Week"
]]

final_merged.head()
```

# Export the Processed Dataset:
The final processed e-commerce dataset is exported as a CSV file for further use and submission.

```python
final_merged.to_csv("processed_ecommerce_dataset.csv", index=False)
```
